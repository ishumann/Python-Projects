```python
import pandas as pd
import torch
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder, StandardScaler
```

```python
df = pd.read_csv(
    "https://raw.githubusercontent.com/gscdit/Breast-Cancer-Detection/refs/heads/master/data.csv"
)
```

```python
df.head()
```

```python
df.drop(columns=["id", "Unnamed: 32"], axis=1, inplace=True)
```

```python
# split the data
X_train, X_test, y_train, y_test = train_test_split(
    df.drop(columns=["diagnosis"]), df["diagnosis"], test_size=0.2
)
```

```python
# scale the data
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

```python
# label encode the target variable
encoder = LabelEncoder()
y_train = encoder.fit_transform(y_train)
y_test = encoder.transform(y_test)
```

```python
# convert to tensors
X_train_tensor = torch.from_numpy(X_train.astype("float32"))
X_test_tensor = torch.from_numpy(X_test.astype("float32"))
y_train_tensor = torch.from_numpy(y_train.astype("float32"))
y_test_tensor = torch.from_numpy(y_test.astype("float32"))
```

```python
# Important Parameters

learning_rate = 0.1
epochs = 25
```

```python
loss_function = nn.BCELoss()
```

```python
# training Pipeline

model = MySimpleNN(X_train_tensor.shape[1])
# model.weight
```

```python
# define the loop
for epoch in range(epochs):

    # forward pass
    y_pred = model(X_train_tensor)

    # loss function
    loss = loss_function(y_pred, y_train_tensor.view(-1, 1))

    # opitmization step with optim

    optimizer = torch.optim.SGD(model.parameters(), lr=learning_rate)

    # zero the gradients
    optimizer.zero_grad()

    # backward pass
    loss.backward()

    # parameters update

    optimizer.step()

    # # manual way to update weights
    # with torch.no_grad():
    #     model.linear.weight -= learning_rate * model.linear.weight.grad
    #     model.linear.bias -= learning_rate * model.linear.bias.grad

    # manual way of zeroing the gradients
    # model.linear.weight.grad.zero_()
    # model.linear.bias.grad.zero_()
    # print loss every epochs
    print(f"Epoch {epoch+1}: loss = {loss.item()}")
```

```python
# model evaluation
with torch.no_grad():
    y_pre = model.forward(X_test_tensor)
    y_pre = torch.where(y_pre > 0.5, 1, 0)

accuracy = (y_pre == y_test_tensor).float().mean()
print(f"Accuracy: {accuracy.item()*100:.2f}%")
```
