```python
import pandas as pd
import torch
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder, StandardScaler
```

```python
df = pd.read_csv(
    "https://raw.githubusercontent.com/gscdit/Breast-Cancer-Detection/refs/heads/master/data.csv"
)
```

```python
df.drop(columns=["id", "Unnamed: 32"], axis=1, inplace=True)
```

```python
# split the data
X_train, X_test, y_train, y_test = train_test_split(
    df.drop(columns=["diagnosis"]), df["diagnosis"], test_size=0.2
)
```

```python
# scale the data
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

```python
# label encode the target variable
encoder = LabelEncoder()
y_train = encoder.fit_transform(y_train)
y_test = encoder.transform(y_test)
```

```python
# convert to tensors
X_train_tensor = torch.from_numpy(X_train)
X_test_tensor = torch.from_numpy(X_test)
y_train_tensor = torch.from_numpy(y_train)
y_test_tensor = torch.from_numpy(y_test)
```

```python
X_train_tensor.shape
```

```python
y_train_tensor.shape
```

```python
# defining the model
```

```python
class MySimpleNN:

    def __init__(self, X):
        self.weights = torch.randn(
            X.shape[1], 1, requires_grad=True, dtype=torch.float64
        )
        self.bias = torch.zeros(1, requires_grad=True, dtype=torch.float64)

    # forward pass
    def forward(self, X):
        z = torch.matmul(X_train_tensor, self.weights) + self.bias
        y_pred = torch.sigmoid(z)
        return y_pred

    # loss function
    def loss_function(self, y_pred, y):
        epsilon = 1e-7
        y_pred = torch.clamp(y_pred, epsilon, 1 - epsilon)

        loss = -(
            y_train_tensor * torch.log(y_pred)
            + (1 - y_train_tensor) * torch.log(1 - y_pred)
        ).mean()
        return loss
```

```python
# Important Parameters

learning_rate = 0.1
epochs = 25
```

```python
# training Pipeline

model = MySimpleNN(X_train_tensor)
model.weights
```

```python
model.bias
```

```python
# define the loop
for epoch in range(epochs):

    # forward pass
    y_pred = model.forward(X_train_tensor)

    # loss function
    loss = model.loss_function(y_pred, y_train_tensor)

    # backward pass
    loss.backward()
    # update weights
    with torch.no_grad():
        model.weights -= learning_rate * model.weights.grad
        model.bias -= learning_rate * model.bias.grad
    # zero the gradients
    model.weights.grad.zero_()
    model.bias.grad.zero_()
    # print loss every epochs
    print(f"Epoch {epoch+1}: loss = {loss.item()}")
```

```python
# model evaluation
with torch.no_grad():
    y_pre = model.forward(X_test_tensor)
    y_pre = torch.where(y_pre > 0.5, 1, 0)

accuracy = (y_pre == y_test_tensor).float().mean()
print(f"Accuracy: {accuracy.item()*100:.2f}%")
```
