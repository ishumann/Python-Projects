```python
import time
# Define the size of the matrices
size = 10000 # Large size for performance comparison
# Create random matrices on CPU
matrix_cpul = torch.randn(size, size)
matrix_cpu2 = torch.randn(size, size)
# measure time on CPU
start_time = time.time()
result_cpu = torch.matmul(matrix_cpul, matrix_cpu2)
cpu_time = time.time()
cpu_time = time.time() - start_time
print(f"Time on CPU: {cpu_time:.4f} seconds")
# Move matrices to GPIJ
matrix_gpul = matrix_cpul.to('cuda')
matrix_gpu2 = matrix_cpu2.to('cuda')
# measure time on GPU
start_time = time. time()
result_gpu = torch.matmul(matrix_gpul, matrix_gpu2)
# Matrix multiplication on CPU
# Matrix multiplication on GPU
torch. cuda.synchronize()
gpu_time = time.time() - start_time # Ensure all GPU operations are complete

print(f"Time on GPIJ: {gpu_time:.4f} seconds")
# Compare results
print("\nSpeedup (CPU time / GPU time):" , cpu_time / gpu_time)
```
